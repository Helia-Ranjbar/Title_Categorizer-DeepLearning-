```python
import pandas as pd
import numpy as np
import keras
```

```python
# Reading/Loading the dataset files
train = pd.read_csv('torob_train.csv')
test = pd.read_csv('torob_test.csv')
train
```

```python
test
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

X = train[['name1', 'name2']]
y = train['cat_id']
y = pd.DataFrame(y, columns=['cat_id'])

# Initialize the OneHotEncoder
encoder = OneHotEncoder(sparse_output=False)

# Fit and transform the column
encoded = encoder.fit_transform(y[['cat_id']])

# Convert to DataFrame with the correct shape
y = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['cat_id']))
```

```python
X
```

```python
y
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
مدل‌سازی
</font>
</h2>

```python
from tensorflow.keras.layers import TextVectorization
WINDOW_LEN = 500

# Initialize TextVectorization layer
vectorizer = TextVectorization(ngrams=2,output_sequence_length=WINDOW_LEN)
vectorizer.adapt(X['name1'])

# Transform the text data into token sequences
X = vectorizer(X['name1']).numpy()

test = vectorizer(test['name1']).numpy()

# Vocabulary size
VOCAB_SIZE = vectorizer.vocabulary_size()

print(f"Vocab size: {VOCAB_SIZE}")

X
```

```python
test
```

```python
from sklearn.model_selection import train_test_split

X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size = 0.1)
```

```python
y_train
```

```python
X_train
```

```python
BATCH_SIZE = 64
EMB_SIZE = 300

model = keras.Sequential()
model.add(keras.layers.Input(batch_shape=(BATCH_SIZE, WINDOW_LEN), dtype='int32'))
model.add(keras.layers.Embedding(VOCAB_SIZE, EMB_SIZE, name='embedding'))
model.add(keras.layers.LSTM(200, return_sequences=True, name='lstm', dropout=0.2))
model.add(keras.layers.GlobalMaxPool1D())
model.add(keras.layers.Dropout(0.2))
model.add(keras.layers.Dense(11, activation='softmax'))

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
epochs = 10
batch_size = 128 
history = model.fit(X_train, y_train, epochs = epochs, batch_size = batch_size, validation_data=(X_valid,y_valid)) # TODO
```
